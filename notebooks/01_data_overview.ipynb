# 导入包

```python
import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 120)
```

# 读取数据

```python
data_path = "../data/raw/marketing_AB.csv"

df = pd.read_csv(data_path)

df.head()
```

# 查看数据基本信息

```python
df.info()
```

# 清洗字段名

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

df.columns
```

# 再次查看前几行

```python
df.head()
```

# 查看数据规模

```python
print("数据行数：", df.shape[0])
print("数据列数：", df.shape[1])
```

# 检查缺失值

```python
df.isnull().sum()
```

# 检查重复用户

```python
duplicate_users = df["user_id"].duplicated().sum()

print("重复 user_id 数量：", duplicate_users)
```

# 检查实验组

```python
df["test_group"].value_counts()
```

# 先做一个最粗的转化率对比

```python
group_summary = df.groupby("test_group").agg(
    users=("user_id", "count"),
    conversions=("converted", "sum"),
    conversion_rate=("converted", "mean")
).reset_index()

group_summary
```
